## Bài 2. Tạo các Tensor thường dùng
Thực hành tạo Tensor ngẫu nhiên, Tensor toàn số 0, toàn số 1 và ma trận đơn vị.

**Yêu cầu:**
- Nhận xét sự khác nhau giữa `torch.rand()` và `torch.randn()`.
- Tạo thêm một Tensor 5 x 5 chứa toàn giá trị 10.


```python
import torch
torch.manual_seed(0)  # de ket qua co the tai lap

a = torch.rand(3, 4)
b = torch.randn(3, 4)
c = torch.zeros(2, 3)
d = torch.ones(2, 3)
e = torch.eye(4)

print("a = torch.rand(3, 4):")
print(a)
print("\nb = torch.randn(3, 4):")
print(b)
print("\nc = torch.zeros(2, 3):")
print(c)
print("\nd = torch.ones(2, 3):")
print(d)
print("\ne = torch.eye(4):")
print(e)

# --- Yeu cau them: Tensor 5x5 toan gia tri 10 ---
f = torch.full((5, 5), 10)
print("\nf = Tensor 5x5 toan gia tri 10:")
print(f)
```

**Nhận xét Bài 2:**
- `torch.rand()`: phân phối **đều (uniform)** trong khoảng [0, 1). Giá trị luôn >= 0 và < 1.
- `torch.randn()`: phân phối **chuẩn (normal)** trung bình 0, phương sai 1, N(0,1). Giá trị có thể âm/dương, thường trong [-3, 3].
- `zeros / ones / eye / full`: tạo Tensor hằng, hữu ích để khởi tạo trọng số, bias, ma trận đơn vị.

## Bài 3. Kiểm tra thông tin Tensor
Sử dụng `shape`, `size()`, `ndim` và `dtype` để kiểm tra các thuộc tính của Tensor.

**Yêu cầu:** Giải thích ý nghĩa của `torch.Size([2, 3, 4])`.


```python
import torch

a = torch.zeros(2, 3, 4)
print("Tensor:")
print(a)
print("Shape:", a.shape)
print("Size:", a.size())
print("Number of dimensions:", a.ndim)
print("Data type:", a.dtype)
print("So phan tu:", a.numel())
```

**Giải thích Bài 3:**
- `torch.Size([2, 3, 4])` nghĩa là Tensor có **3 chiều** (ndim=3):
  - Chiều 0 có độ dài 2 (2 khối),
  - Chiều 1 có độ dài 3 (mỗi khối 3 hàng),
  - Chiều 2 có độ dài 4 (mỗi hàng 4 cột).
- Tổng số phần tử = 2 x 3 x 4 = 24.
- `shape` và `size()` trả về cùng giá trị; `shape` là thuộc tính, `size()` là phương thức.

## Bài 4. Slicing và Indexing
Truy xuất một phần dữ liệu từ Tensor nhiều chiều.

**Yêu cầu:**
- Cho biết shape của `a[1]` và `a[1:, 2:4]`.
- Thử lấy hai phần tử cuối cùng trên chiều thứ ba.


```python
import torch
torch.manual_seed(0)

a = torch.randn(3, 4, 5)
print("Original tensor:")
print(a)
print("shape cua a:", a.shape)

print("\nTensor a[1]:")
print(a[1])
print("shape cua a[1]:", a[1].shape)

print("\nTensor a[1:, 2:4]:")
print(a[1:, 2:4])
print("shape cua a[1:, 2:4]:", a[1:, 2:4].shape)

# --- Lay hai phan tu cuoi cung tren chieu thu ba (dim=2) ---
last_two = a[:, :, -2:]
print("\nHai phan tu cuoi tren chieu thu ba a[:, :, -2:]:")
print(last_two)
print("shape:", last_two.shape)
```

**Nhận xét Bài 4:**
- `a` có shape (3, 4, 5).
- `a[1]` lấy lát cắt tại index 1 của chiều 0 -> shape (4, 5) (mất 1 chiều).
- `a[1:, 2:4]` lấy chiều 0 từ 1 đến hết (2 phần tử), chiều 1 lấy index 2,3 (2 phần tử), giữ nguyên chiều 2 (5 phần tử) -> shape (2, 2, 5).
- Để lấy 2 phần tử cuối trên chiều thứ ba: `a[:, :, -2:]` -> shape (3, 4, 2).

## Bài 5. Reshape và Flatten
Thay đổi hình dạng Tensor mà không làm thay đổi số phần tử.

**Yêu cầu:**
- Kiểm tra số phần tử trước và sau khi reshape.
- Thử reshape Tensor thành kích thước 2 x 6.


```python
import torch
torch.manual_seed(0)

a = torch.randn(3, 4)
print("Original:")
print(a)
print("shape:", a.shape, "| numel:", a.numel())

print("\nFlatten:")
print(a.ravel())
print("shape flatten:", a.ravel().shape)

print("\nReshape to 3 x 2 x 2:")
b = a.reshape(3, 2, 2)
print(b)
print("shape:", b.shape, "| numel:", b.numel())

# --- Yeu cau them ---
print("\nReshape to 2 x 6:")
c = a.reshape(2, 6)
print(c)
print("shape:", c.shape, "| numel:", c.numel())

assert a.numel() == b.numel() == c.numel() == 12, "reshape phai giu nguyen so phan tu!"
```

**Nhận xét Bài 5:**
- Tensor gốc 3 x 4 = 12 phần tử.
- `ravel()` / `flatten()` trải phẳng thành vector 12 phần tử.
- `reshape(3, 2, 2)` và `reshape(2, 6)` đều có 12 phần tử — reshape chỉ đổi hình dạng, không đổi dữ liệu/tổng số phần tử.
- Điều kiện reshape hợp lệ: tích các kích thước mới phải bằng `numel` gốc.

## Bài 6. Phép toán và nhân ma trận
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử `@`.

**Yêu cầu:**
- Phân biệt phép toán `a * b` với `a @ b.T`.
- Cho biết shape của kết quả `a @ b.T`.


```python
import torch
torch.manual_seed(0)

a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\na * b (nhan tung phan tu):")
print(a * b)
print("shape:", (a * b).shape)

print("\nMatrix multiplication a @ b.T:")
print(a @ b.T)
print("shape:", (a @ b.T).shape)
```

**Nhận xét Bài 6:**
- `a * b` là **nhân từng phần tử (element-wise)**: yêu cầu cùng shape (hoặc broadcast được). Với `a, b: (2,3)` -> kết quả `(2, 3)`.
- `a @ b.T` là **nhân ma trận**: `a:(2,3)` nhân `b.T:(3,2)` -> kết quả `(2, 2)`. Mỗi phần tử là tích vô hướng của một hàng `a` với một hàng `b`.
- Công thức shape: (m x n) @ (n x p) -> (m x p). Ở đây (2x3)@(3x2)->(2x2).

## Bài 7. Các hàm thống kê trên Tensor
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

**Yêu cầu:** Thay `dim=0` bằng `dim=1` và nhận xét sự khác nhau.


```python
import torch
torch.manual_seed(0)

a = torch.randn(3, 4)
print(a)

print("\nMean (dim=0, theo cot):")
print(torch.mean(a, dim=0))

print("\nStandard deviation (dim=0):")
print(torch.std(a, dim=0))

print("\nCumulative sum (dim=0):")
print(torch.cumsum(a, dim=0))

print("\n===== So sanh voi dim=1 (theo hang) =====")
print("\nMean (dim=1):")
print(torch.mean(a, dim=1))

print("\nStandard deviation (dim=1):")
print(torch.std(a, dim=1))

print("\nCumulative sum (dim=1):")
print(torch.cumsum(a, dim=1))
```

**Nhận xét Bài 7:**
- Với Tensor shape (3, 4) (3 hàng, 4 cột):
  - `dim=0`: tính dọc theo các **hàng** -> gom theo **cột**. Kết quả shape (4,) (1 giá trị cho mỗi cột). `cumsum` cộng dồn từ trên xuống dưới.
  - `dim=1`: tính dọc theo các **cột** -> gom theo **hàng**. Kết quả shape (3,) (1 giá trị cho mỗi hàng). `cumsum` cộng dồn từ trái sang phải.
- Quy tắc nhớ: `dim=k` là chiều bị triệt tiêu (collapse) sau khi tính mean/std.

## Bài 8. Autograd cơ bản
Tạo biến có `requires_grad=True` và dùng `backward()` để tính đạo hàm tự động.

**Yêu cầu:**
- Tính thủ công đạo hàm của y = x^2 tại x = 3.6 và so sánh với `x.grad`.
- Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.


```python
import torch

x = torch.tensor(
    3.6,
    requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

# --- Tinh thu cong: y = x^2 -> dy/dx = 2x = 2*3.6 = 7.2 ---
print("Tinh thu cong dy/dx = 2*3.6 =", 2*3.6)

# --- Yeu cau them: y = 3*x**2 + 2*x + 1 ---
x2 = torch.tensor(3.6, requires_grad=True)
y2 = 3*x2**2 + 2*x2 + 1
y2.backward()
print("\nVoi y = 3*x**2 + 2*x + 1 tai x=3.6:")
print("y2 =", y2.item())
print("x2.grad (autograd) =", x2.grad.item())
print("Tinh thu cong dy/dx = 6x+2 =", 6*3.6+2)
```

**Nhận xét Bài 8:**
- y = x^2 => dy/dx = 2x. Tại x=3.6: 2*3.6=7.2 khớp với `x.grad`.
- y = 3x^2+2x+1 => dy/dx = 6x+2. Tại x=3.6: 6*3.6+2=23.6 khớp với autograd.
- `requires_grad=True` yêu cầu PyTorch lưu đồ thị tính toán; `backward()` lan truyền ngược để tính gradient.

## Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức y = x^2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

**Yêu cầu:**
- So sánh ba hệ số học được với `[1, 2, 3]`.
- Giải thích vai trò của `optimizer.zero_grad()`, `mse.backward()` và `optimizer.step()`.


```python
import numpy as np
import torch
np.random.seed(0)
torch.manual_seed(0)

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
print("He so thuc: [1, 2, 3]")
```

**Nhận xét Bài 9:**
- Ba hệ số học được xấp xỉ [1, 2, 3] (sai số nhỏ do làm tròn / số epoch hữu hạn). Mô hình hồi quy tuyến tính trên đặc trưng [x^2, x, 1] đã khôi phục đúng đa thức sinh dữ liệu.
- Vai trò 3 lệnh trong vòng lặp:
  - `optimizer.zero_grad()`: xóa gradient cũ (PyTorch cộng dồn gradient, nên phải reset mỗi epoch).
  - `mse.backward()`: tính gradient của loss theo `w` bằng autograd (lan truyền ngược).
  - `optimizer.step()`: cập nhật `w` theo hướng ngược gradient (NAdam là biến thể Gradient Descent có momentum + lr thích nghi).

## Bài 10. Giải hệ phương trình bằng Autograd
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

**Yêu cầu:**
- Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.
- Nhận xét vì sao bài toán có thể có nhiều nghiệm.


```python
import random
import torch
random.seed(0)
torch.manual_seed(0)

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)

# --- Kiem tra nghiem ---
print("\nKiem tra:")
print(f"A+B = {(A+B).item():.4f} (mong doi 9)")
print(f"C-D = {(C-D).item():.4f} (mong doi 1)")
print(f"A+C = {(A+C).item():.4f} (mong doi 8)")
print(f"B-D = {(B-D).item():.4f} (mong doi 2)")
```

**Nhận xét Bài 10:**
- Nghiệm thỏa mãn cả 4 phương trình với sai số ~0 (ví dụ A~3, B~6, C~5, D~4 tùy lần chạy — học viên đối chiếu số in ra).
- Bài toán **có thể có nhiều nghiệm** vì 4 phương trình không độc lập tuyến tính: (A+B)-(A+C)+(C-D)-(B-D) = 0恒成立 với mọi A,B,C,D, nên hạng của hệ < 4. Hệ phụ thuộc tuyến tính nên có 1 bậc tự do -> vô số nghiệm. Gradient descent sẽ hội tụ về một nghiệm gần điểm khởi tạo ngẫu nhiên.

#### Bài 11.4. Gradient Descent với y = 2x^2 - 3x + 5


```python
import numpy as np
import torch
np.random.seed(1)
torch.manual_seed(1)

polynomial = np.poly1d([2, -3, 5])  # y = 2x^2 - 3x + 5
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([X*X, X, np.ones_like(X)])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients (mong doi ~[2, -3, 5]):")
print(w)
```
